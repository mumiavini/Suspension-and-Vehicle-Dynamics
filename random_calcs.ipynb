# Parâmetros Iniciais de Projeto — FSAE26 (PUCPR Racing #27)

## 1. Resistência a capotamento (rollover)

**Critério:** o carro não deve capotar antes de derrapar.

$$\frac{T}{2 \cdot h_{CG}} > \tan(60°) \approx 1{,}73$$

Ou seja: **T > 3,46 · h_CG**.

Com bitola dianteira 1200 mm e h_CG estimado em ~289 mm: 1200 / (2 × 289) = 2,08 > 1,73 ✓

> 60° é o ângulo lateral máximo antes de capotamento estático. Na prática,
> FSAE nunca chega perto — o pneu perde aderência bem antes. O critério garante margem
> mesmo em cenários de curva + lombada.

## 2. Layout básico

Relação alta (>0,85) favorece estabilidade lateral mas deixa o carro "preguiçoso" em transições.
Relação baixa (<0,70) torna o carro nervoso e instável em regime permanente.
A faixa 0,75–0,80 é consenso em FSAE (Milliken RCVD Ch. 8; dados competitivos de Michigan/Lincoln).

## 3. Altura do centro de gravidade (h_CG)

### Fórmula — média ponderada das massas

$$h_{CG} = \frac{\sum_i m_i \cdot h_i}{\sum_i m_i} = \frac{m_1 h_1 + m_2 h_2 + \cdots + m_n h_n}{m_{total}}$$

### Estimativa por subsistema

| Subsistema | Massa (kg) | Altura CG (mm) | m·h (kg·mm) |
|---|---|---|---|
| Piloto | 70 | 325 | 22 750 |
| Motor / powertrain | 62 | 275 | 17 050 |
| Chassi + carroceria | 32 | 315 | 10 080 |
| Não-suspenso (4 cantos) | 48 | 240 | 11 520 |
| Combustível, sistemas | 18 | 280 | 5 040 |

**Totais:** m_total ≈ 230 kg, Σ(m·h) ≈ 66 440 kg·mm

$$h_{CG} \approx \frac{66\,440}{230} \approx 289 \text{ mm}$$

> Estimativa de engenharia — validar com teste de inclinação (tilt test).
> Fórmula do tilt test: $h_{CG} = \frac{\Delta W \cdot T}{W_{total} \cdot \sin\theta}$.
> Precisão típica: ±5–10 mm. Um erro de 10 mm muda a transferência de carga lateral em ~3%.

## 4. Roll centre height (h_RC)

| Eixo | Faixa recomendada | Valor de projeto |
|---|---|---|
| Dianteiro | 25–45 mm | 35 mm |
| Traseiro | 45–65 mm | 55 mm |
| Limite superior | < 70 mm em ambos | — |

**Traseiro mais alto que dianteiro** — inclina o eixo de rolagem para baixo na frente. Convenção estabelecida: dá estabilidade direcional e direciona mais transferência de carga geométrica para o eixo traseiro.

### Por que esses valores?

A transferência de carga lateral total ($\Delta F_z$) se decompõe em duas parcelas:

- **Geométrica (via RC):** transferência instantânea, não passa pelas molas — o engenheiro **não controla** depois de definida a geometria.
- **Elástica (via molas/ARB):** transferência controlável por troca de molas e ajuste de ARB.

O objetivo é maximizar a parcela elástica, porque ela é ajustável.

**RC baixo (25–45 mm dianteiro):**
- Maior parcela elástica → **mais controle** sobre o balanço do carro.
- Menor **força de jacking** — a componente vertical da reação lateral no pneu. Jacking levanta o chassi na curva, consome curso de suspensão e piora o controle de camber.
- Menor migração do RC com bump → geometria mais previsível em transientes.

**RC alto demais (> 70 mm):**
- Transferência predominantemente geométrica → o engenheiro perde controle do balanço F/R.
- Força de jacking elevada levanta o CG efetivo na curva.
- IC fica curto (FVSA pequeno) → grande variação de camber por mm de curso, difícil de controlar.

**RC muito baixo ou negativo:**
- Quase toda a transferência é elástica → exige molas muito rígidas ou ARB pesada, o que degrada ride quality e compliance.
- ARB excessivamente rígida acopla os dois lados, degradando resposta em obstáculo de uma roda.

> Fonte: Milliken RCVD Ch. 17 (construção gráfica); Optimum G Tech Tips.
> Regra prática: priorize **baixa migração** em rolagem sobre o valor absoluto.

## 5. FVSA — Front-View Swing Arm length

| Eixo | Faixa recomendada | Valor de projeto |
|---|---|---|
| Dianteiro | 1400–1800 mm | 1500 mm |
| Traseiro | 1500–2000 mm | 1600 mm |

### O que é o FVSA?

Distância do centro da roda até o **instant centre (IC)** na vista frontal.
Determina a taxa de ganho de camber em bump:

$$\text{camber gain} \approx \frac{-57{,}3}{\text{FVSA}} \text{ [°/mm]}$$

Com FVSA = 1500 mm → ganho ≈ **-0,038 °/mm** → em 25 mm de bump: **-0,95°** de camber (negativo = topo para dentro = favorável).

### Verificação de fechamento com rolagem

Em 1,5 g lateral com roll gradient de 1,0 °/g → rolagem de 1,5°.

A roda externa sobe ~25 mm. Ganho de camber: -0,95°. Mas a rolagem do chassi adiciona +1,5° ao camber da roda externa.

Resultado líquido: +1,5° − 0,95° = **+0,55° de camber** em relação ao solo → é isso que o **camber estático negativo** (-1,0° a -1,5°) compensa. O pneu fica com camber negativo líquido na curva.

### Por que esses valores?

**FVSA curto (< 1200 mm):**
- Ganho de camber alto, mas o IC fica muito perto da roda.
- RC migra muito com o curso → geometria **instável**, difícil de prever em transientes.
- Scrub excessivo (arraste lateral do pneu no contato com o solo).

**FVSA longo (> 2500 mm):**
- Ganho de camber insuficiente — o pneu não acompanha a rolagem do chassi.
- Em 25 mm de bump com FVSA = 3000 mm: ganho = -0,48° — pode não compensar a rolagem.

**FVSA 1400–1800 mm (sweet spot):**
- Ganho de camber suficiente para compensar ~60–80% da rolagem nos primeiros 25 mm de curso.
- RC com migração moderada e previsível.
- Scrub aceitável.
- Traseiro ligeiramente mais longo: menor sensibilidade à transferência de carga traseira, prioriza estabilidade.

> Fonte: Milliken RCVD Ch. 17 (construção do IC); primeira ordem confirmada pela
> relação camber_gain ≈ -1/FVSA (Milliken eq. 17.x).

## 6. Roll gradient

**Faixa FSAE típica: 0,8 a 1,5 °/g**

### Fórmula

$$\frac{\phi}{A_y} = \frac{m_s \cdot g \cdot d}{K_\phi - m_s \cdot g \cdot d}$$

onde:
- $m_s$ = massa suspensa [kg]
- $d = h_{CG} - h_{eixo\ de\ rolagem}$ = braço de momento de rolagem [mm → converter para m]
- $K_\phi$ = rigidez total de rolagem (molas + ARB, frente + trás) [Nm/rad]
- $\phi / A_y$ = roll gradient [rad/g] → converter para °/g

### Rigidez de rolagem necessária

Invertendo a fórmula:

$$K_\phi = m_s \cdot g \cdot d \cdot \left(\frac{1}{\phi_{alvo}} + 1\right)$$

onde $\phi_{alvo}$ é o roll gradient em **rad/g**.

### Por que 0,8 a 1,5 °/g?

**Abaixo de 0,8 °/g (carro muito rígido em rolagem):**
- Requer molas e/ou ARB muito rígidas → **ride quality degradado**, perda de compliance sobre curbing.
- Ganho de camber por rolagem mínimo — desperdiça o projeto de FVSA (projetou-se ganho que nunca é ativado).
- Pneu FSAE (10" bias-ply ou radial de baixo perfil) não exige platô de camber tão estreito — não é F1 com slick de alta sensibilidade a camber.
- Maior transferência de carga dinâmica (pico), reduzindo o grip total do eixo.

**Acima de 1,5 °/g (carro mole em rolagem):**
- Rolagem excessiva → camber positivo na roda externa (perda de grip lateral).
- Curso de suspensão consumido por rolagem, restando pouco para bumps e curbing.
- Impressão subjetiva de "carro mole" — prejudica confiança do piloto em transições rápidas (slalom, chicane).
- Transferência de carga mais lenta (predominantemente elástica), mas o delay percebido pode prejudicar feeling de entrada.

**Faixa 0,8–1,5 °/g (compromisso ideal):**
- Permite que o FVSA compense 60–80% da rolagem, mantendo camber negativo na roda externa.
- Ride quality aceitável para curbing e lombadas típicas de FSAE.
- A maioria dos carros competitivos de Michigan/Lincoln opera nessa faixa.
- Permite projetar a distribuição F/R de rigidez de rolagem (via ARB) para ajustar balanço — espaço de ajuste real, não teórico.

> Fonte: Milliken RCVD Ch. 16 (roll gradient com correção pendular);
> dados competitivos FSAE (Design Event reports, Optimum G seminars).

### Rigidez torcional do chassi

A rigidez torcional do chassi deve ser **3 a 5× a rigidez total de rolagem** ($K_\phi$).

Abaixo de 3×, o chassi se torna o "spring rate limitante" e o comportamento fica imprevisível.
